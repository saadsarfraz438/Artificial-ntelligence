#Visualization 1 — Histogram
###Sleep Hours Distribution

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df['Average Sleep Hours Per Night'], bins=10)
plt.title('Distribution of Sleep Hours')
plt.xlabel('Sleep Hours')
plt.ylabel('Frequency')
plt.show()
```

#Visualization 2 — Bar Chart
###Stress Level Distribution

```python
import seaborn as sns

plt.figure(figsize=(8,5))
sns.countplot(x='Average Stress Level', data=df)
plt.title('Stress Level Distribution')
plt.xlabel('Stress Level')
plt.ylabel('Count')
plt.show()
```

#Visualization 3 — Pie Chart
###Migraine vs Non-Migraine

```python
migraine_counts = df['Do you experience migraines? 9(Y)'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(migraine_counts,
        labels=['No Migraine', 'Migraine'],
        autopct='%1.1f%%')
plt.title('Migraine Occurrence')
plt.show()
```

#Visualization 4 — Scatter Plot
###Screen Exposure vs Sleep Hours

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x='Average Screen Exposure (per day)',
    y='Average Sleep Hours Per Night',
    hue='Do you experience migraines? 9(Y)',
    data=df
)

plt.title('Screen Exposure vs Sleep Hours')
plt.xlabel('Screen Exposure')
plt.ylabel('Sleep Hours')
plt.show()
```

#Visualization 5 — Correlation Heatmap


```python
plt.figure(figsize=(10,8))

sns.heatmap(
    df.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Heatmap')
plt.show()
```
